```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import TimeSeriesSplit, GridSearchCV
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from xgboost import XGBRegressor

sns.set_style("whitegrid")
np.random.seed(42)
try:
    from google.colab import files
    uploaded = files.upload()
    path = list(uploaded.keys())[0]
except ImportError:
    path = "debutanizer_data.csv"

df = pd.read_csv(path, index_col=0)
print(df.shape)
df.head()
df = df.reset_index(drop=True)
df.index.name = "t"

df.info()
print()
print("missing values per column:")
print(df.isna().sum())
print("duplicate rows:", df.duplicated().sum())
df.describe()

rows = []
for c in df.columns:
    flat_run = (df[c] != df[c].shift()).cumsum().value_counts().max()
    q1, q3 = df[c].quantile([0.25, 0.75])
    iqr = q3 - q1
    n_out = ((df[c] < q1 - 1.5 * iqr) | (df[c] > q3 + 1.5 * iqr)).sum()
    rows.append([c, flat_run, n_out])

checks = pd.DataFrame(rows, columns=["column", "longestflatrun", "iqroutliers"])
print(checks)
x_cols = ["x1", "x2", "x3", "x4", "x5", "x6", "x7"]

fig, axes = plt.subplots(8, 1, figsize=(12, 15))
for ax, c in zip(axes, ["y"] + x_cols):
    ax.plot(df.index, df[c], linewidth=0.8)
    ax.set_ylabel(c)
axes[0].set_title("All variables over time (y = butane concentration)")
axes[-1].set_xlabel("time step")
plt.tight_layout()
plt.show()
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].hist(df["y"], bins=40, color="tab:red", edgecolor="white")
ax[0].set_title("Distribution of butane concentration (y)")
ax[0].set_xlabel("y ")

sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0, ax=ax[1])
ax[1].set_title("Correlation at the same time step")
plt.tight_layout()
plt.show()
max_lag = 80
lag_corr = pd.DataFrame(index=range(max_lag + 1), columns=x_cols, dtype=float)
for c in x_cols:
    for k in range(max_lag + 1):
        lag_corr.loc[k, c] = df["y"].corr(df[c].shift(k))

lag_corr.plot(figsize=(11, 4.5))
plt.axhline(0, color="black", linewidth=0.7)
plt.xlabel("lag k (steps): correlation of y(t) with x(t-k)")
plt.ylabel("correlation")
plt.title("Delayed inputs explain y much better than current ones")
plt.show()

best = pd.DataFrame({
    "lag at strongest correlation": lag_corr.abs().idxmax(),
    "correlation there": [lag_corr.loc[lag_corr[c].abs().idxmax(), c] for c in x_cols],
    "correlation at lag 0": lag_corr.loc[0],
}).round(2)
print(best)
lags = [0, 5, 10, 15, 20, 25, 30]

feat_A = pd.DataFrame(index=df.index)
for c in x_cols:
    for k in lags:
        feat_A[f"{c}_lag{k}"] = df[c].shift(k)
print(feat_A.head())

feat_B = feat_A.copy()
for k in [1, 2, 3, 4]:
    feat_B[f"ylag{k}"] = df["y"].shift(k)

valid = feat_B.dropna().index
X_A = feat_A.loc[valid]
X_B = feat_B.loc[valid]
y = df.loc[valid, "y"]

print("rows left:", len(y), "(dropped", len(df) - len(y), ")")
print("features in A:", X_A.shape[1], "| features in B:", X_B.shape[1])
split = int(len(y) * 0.7)

XA_train, XA_test = X_A.iloc[:split], X_A.iloc[split:]
XB_train, XB_test = X_B.iloc[:split], X_B.iloc[split:]
y_train, y_test = y.iloc[:split], y.iloc[split:]

print("train:", len(y_train), "rows | test:", len(y_test), "rows")
tscv = TimeSeriesSplit(n_splits=4)

searches = {
    "Ridge": GridSearchCV(
        make_pipeline(StandardScaler(), Ridge()),
        {"ridge__alpha": [0.01, 0.1, 1, 10, 100, 1000]},
        cv=tscv, scoring="r2"),
    "Random Forest": GridSearchCV(
        RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1),
        {"max_depth": [4, 8, None], "min_samples_leaf": [1, 5, 20]},
        cv=tscv, scoring="r2"),
    "XGBoost": GridSearchCV(
        XGBRegressor(n_estimators=300, subsample=0.8, colsample_bytree=0.8,
                     random_state=42, verbosity=0, n_jobs=-1),
        {"max_depth": [2, 3, 4], "learning_rate": [0.03, 0.1]},
        cv=tscv, scoring="r2"),
}

def run_version(X_train, X_test, label):
    results, preds, fitted = [], {}, {}
    for name, search in searches.items():
        search.fit(X_train, y_train)
        model = search.best_estimator_
        p = model.predict(X_test)
        results.append({
            "version": label, "model": name,
            "best CV R2": round(search.best_score_, 3),
            "test R2": round(r2_score(y_test, p), 3),
            "test RMSE": round(np.sqrt(mean_squared_error(y_test, p)), 4),
            "test MAE": round(mean_absolute_error(y_test, p), 4),
        })
        preds[name] = p
        fitted[name] = model
        print(label, name, "done, best params:", search.best_params_)
    return results, preds, fitted

res_A, pred_A, model_A = run_version(XA_train, XA_test, "A (no past y)")
res_B, pred_B, model_B = run_version(XB_train, XB_test, "B (with past y)")
results = pd.DataFrame(res_A + res_B)
print(results)
mean_pred = np.full(len(y_test), y_train.mean())
print("always-predict-the-mean baseline: R2 = %.3f, RMSE = %.4f"
      % (r2_score(y_test, mean_pred), np.sqrt(mean_squared_error(y_test, mean_pred))))

fig, ax = plt.subplots(figsize=(8, 4.5))
width = 0.35
pos = np.arange(3)
names = ["Ridge", "Random Forest", "XGBoost"]
ax.bar(pos - width/2, [r["test R2"] for r in res_A], width, label="A: no past y", color="tab:blue")
ax.bar(pos + width/2, [r["test R2"] for r in res_B], width, label="B: with past y", color="tab:orange")
ax.set_xticks(pos)
ax.set_xticklabels(names)
ax.axhline(0, color="black", linewidth=0.8)
ax.set_ylabel("test R2")
ax.set_title("Test R2 by model and version")
ax.legend()
plt.show()
best_A = results[results.version.str.startswith("A")].sort_values("test R2").iloc[-1]["model"]
best_B = results[results.version.str.startswith("B")].sort_values("test R2").iloc[-1]["model"]
print("best model A:", best_A, "| best model B:", best_B)

t_test = y_test.index
fig, axes = plt.subplots(2, 1, figsize=(12, 8), sharex=True)

axes[0].plot(t_test, y_test, color="black", label="actual", linewidth=1.2)
for name in names:
    axes[0].plot(t_test, pred_A[name], label=name, linewidth=0.9, alpha=0.8)
axes[0].set_title("Version A (no past y): actual vs predicted on the test period")
axes[0].set_ylabel("butane (scaled)")
axes[0].legend(ncol=4)

axes[1].plot(t_test, y_test, color="black", label="actual", linewidth=1.2)
for name in names:
    axes[1].plot(t_test, pred_B[name], label=name, linewidth=0.9, alpha=0.8)
axes[1].set_title("Version B (with past y)")
axes[1].set_ylabel("butane (scaled)")
axes[1].set_xlabel("time step")
plt.tight_layout()
plt.show()
print(results[["version", "model", "best CV R2", "test R2"]].to_string(index=False))
```
